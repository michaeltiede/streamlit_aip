```python
# ... your existing imports and data loading ...
def check_county(df, state_name, county_name_partial):
    county_check = df[(df['State'] == state_name) & (df['County'].str.contains(county_name_partial))]
    if county_check.empty:
        print(f"{county_name_partial} County not found in {state_name}!")
        return pd.DataFrame()
    county_check['County_name'] = county_check['County'].str.replace(r'<.*?>', '', regex=True)
    return county_check[['County', 'County_name', 'Latitude', 'Longitude', 'FIPS']]

# Check Oglala Lakota County
check_county(df, 'South Dakota', 'Oglala Lakota')


```
